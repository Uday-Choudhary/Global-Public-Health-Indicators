###  Data Cleaning Steps

- Filled missing **numerical values** using the **median** to reduce the effect of outliers.  
- Filled missing **categorical values** using the **mode** (most frequent value).  
- Checked and removed any duplicate rows to avoid biased analysis.  
- Verified and corrected data types to ensure accurate calculations and visualizations.  

These steps prepared the dataset for reliable analysis and modeling.


```python
df.head()
```

```python
df.isnull().sum()
```

```python
num_cols = df.select_dtypes(include="number").columns
df[num_cols] = df[num_cols].fillna(df[num_cols].median())

cat_cols = df.select_dtypes(include='object').columns
df[cat_cols] = df[cat_cols].fillna(df[cat_cols].mode().iloc[0])
```

```python
df.drop_duplicates(inplace=True)
```

### Analysis of Health Outcome Trends: Developed vs Developing Countries

- Grouped countries by **Status** (Developed / Developing) and **Year**.  
- Analyzed key health indicators such as **Life Expectancy** and **Infant Mortality** over time.  
- Used **data mining techniques** to calculate trends and improvement rates for each group.  
- Visualized results with **line plots** to compare progress between developed and developing countries.  
- Determined whether the health outcome gap is **narrowing, widening, or stable**, providing insights into global health disparities.


```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(14, 10))
corr = df.select_dtypes(include='number').corr()

sns.heatmap(corr, annot=True, linewidths=0.5)
plt.title("Correlation Heatmap of Life Expectancy Dataset")
plt.show()
```

The heatmap visually represents the correlation between all numeric variables in the dataset, allowing us to identify strong positive or negative relationships that influence life expectancy.

| **Correlation Value** | **Meaning**                     |

```python
developed = df[df['Status'] == 'Developed']
developing = df[df['Status'] == 'Developing']
```

```python
plt.figure(figsize=(12,6))

sns.lineplot(data=developed, x='Year', y='infant deaths', label='Developed')
sns.lineplot(data=developing, x='Year', y='infant deaths', label='Developing')

plt.title("Infant Mortality Trends")
plt.show()
```

```python
life_dev = developed.groupby("Year")["Life expectancy "].mean()
life_devlp = developing.groupby("Year")["Life expectancy "].mean()

rate_dev = life_dev.pct_change().mean()
rate_devlp = life_devlp.pct_change().mean()

print("Improvement Rate (Developed):", rate_dev)
print("Improvement Rate (Developing):", rate_devlp)
```

```python
gap = life_dev - life_devlp
gap.plot(figsize=(10,6), title="Life Expectancy GAP (Developed - Developing)")
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
features = [
    'Life expectancy ',
    'Adult Mortality',
    'infant deaths',
    'Income composition of resources',
    'Schooling',
    'percentage expenditure',
    ' HIV/AIDS',
    ' thinness 5-9 years',
    'GDP'
]
data = df[features].dropna()
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data)
```

```python
sse = []
K = range(2, 10)

for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(scaled_data)
    sse.append(kmeans.inertia_)

plt.plot(K, sse, marker='o')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('SSE')
plt.title('Elbow Method to Determine Optimal k')
plt.show()
```

```python
kmeans = KMeans(n_clusters=2, random_state=42)
clusters = kmeans.fit_predict(scaled_data)
df['Cluster'] = clusters
```

```python
pd.crosstab(df['Status'], df['Cluster'])
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
reduced = pca.fit_transform(scaled_data)

df['PC1'] = reduced[:, 0]
df['PC2'] = reduced[:, 1]

plt.figure(figsize=(9, 6))
sns.scatterplot(data=df, x='PC1', y='PC2', hue='Cluster', palette='viridis')
plt.title("K-Means Clustering of Countries (PCA Reduced)")
plt.show()
```
